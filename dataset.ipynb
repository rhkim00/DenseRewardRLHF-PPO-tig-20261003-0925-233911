```python
from datasets import load_dataset

ds_stream = load_dataset(
    "deepmind/code_contests",
    split="train",
    streaming=True
)

# 앞에서 100개만 가져오기
small_samples = []
for i, ex in enumerate(ds_stream):
    if i >= 100:
        break
    small_samples.append(ex)
```

```python
from datasets import Dataset

small_ds = Dataset.from_list(small_samples)
small_ds.to_json("code_contests_train_100.jsonl")
```
